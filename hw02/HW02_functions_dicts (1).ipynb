````python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
````

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База

# 1
def to_num(s):
    """Строку вида '15 990' или '8070.0' превращает в число."""
    return float(s.replace(" ", ""))

def clean_city(s):
  return s.strip().title()

# 2
categories_count = {}
for row in rows:
  category = row["category"]
  categories_count[category] = categories_count.get(category, 0) + 1

for c , v in sorted(categories_count.items(), key=lambda t: t[1], reverse=True):
    print(f"{c:<10} {v:>7,}")

print("")
# 3
paid_categories = {}
for row in rows:
  if row["status"] == "оплачен":
    category = row["category"]
    paid_categories[category] = paid_categories.get(category, 0) + to_num(row["amount"])

for c , v in sorted(paid_categories.items(), key=lambda t: t[1], reverse=True):
    print(f"{c:<10} {v:>7,} ₸")
print("")
# 4
months = {}
for row in rows:
  month = row["order_date"][:7]
  months[month] = months.get(month, 0) + 1

for c , v in sorted(months.items(), key=lambda t: t[1], reverse=True):
    print(f"{c:<10} {v:>7,}")
````
