## 1. Series: массив NumPy с ярлыками (02-Series)
Series хранит значения как массив NumPy и добавляет к каждому значению **ярлык** (index). По ярлыку можно достать значение, как из словаря.

````python
%pip install numpy
%pip install pandas
````

````python
import numpy as np
import pandas as pd

revenue = pd.Series([120, 45, 80], index=["Алматы", "Шымкент", "Астана"])
print(revenue)
print(revenue["Шымкент"])     # по ярлыку, как в словаре
print(revenue.values)          # внутри обычный массив NumPy
print(revenue * 1000)          # операции как с массивом
````

## 5. Строки: loc и iloc

| | Что берёт | Пример | Срез |
|---|---|---|---|
| `loc` | по **ярлыку** индекса и имени столбца | `df.loc[1001, "amount"]` | включает правую границу |
| `iloc` | по **позиции**, как в списке | `df.iloc[0, 5]` | правая граница не входит |

Сейчас ярлыки строк совпадают с позициями: 0, 1, 2, ... Дальше увидим, что после фильтра это уже не так.

````python
print(df.loc[1001])                         # строка с ярлыком 1001
print('---------')
print(df.loc[1001, "amount"])    
print('---------')# одна клетка
print(df.iloc[0, 5])                        # строка 0, столбец 5 (amount)
````

````python
df.loc[1001]
````

````python
# ПРЕДСКАЖИТЕ: сколько строк в каждом срезе?
print(len(df.loc[0:2]), len(df.iloc[0:2]))
````

````python
df.loc[0:2]
````

````python
# Несколько строк и столбцов сразу: списки ярлыков
df.loc[[1001, 2268, 2966], ["order_date", "customer_id", "amount"]]
````

## 6. Фильтры: маска, как в NumPy
Условие над столбцом даёт Series из `True/False`. Таблица в квадратных скобках оставляет строки, где `True`.

| Нужно | Код |
|---|---|
| и | `(условие1) & (условие2)` |
| или | `(условие1) \| (условие2)` |
| не | `~(условие)` |
| одно из значений | `df["city"].isin(["Шымкент", "шымкент"])` |

**Скобки обязательны:** `&` выполняется раньше `>` и `==` (вопрос 3 разминки).

````python
df["customer_id"] == "TEST"
````

````python
test = df[df["customer_id"] == "TEST"]
test
````

````python
test[["order_date", "city", "amount", "status"]]
````

````python
df[df["customer_id"] == "TEST"][["order_date", "city", "amount", "status"]]
````

````python
# Строки и столбец за одно действие: loc[маска, столбец]
big = df.loc[(df["amount"] > 100_000) & (df["customer_id"] != "TEST"), "amount"]
big
````

````python
print(len(big), big.sum())
````

### Индекс после фильтра
Фильтр сохраняет **старые ярлыки** строк. Поэтому `loc[0]` может не найти строку, а `iloc[0]` всегда берёт первую по порядку.

````python
df
````

````python
sh
````

````python
# ПРЕДСКАЖИТЕ: что сработает, а что упадёт?
print(sh.index[:5].tolist())
print(sh.iloc[0]["city"])
try:
    print(sh.loc[0])
except KeyError as e:
    print("KeyError:", e)
````

> Связь с уроком 3: срез NumPy – вид на те же данные. Если после фильтра вы будете **менять** результат, сделайте копию: `sh = df[маска].copy()`. Иначе Pandas выдаст предупреждение `SettingWithCopyWarning` (в новых версиях – молча не изменит `df`).
